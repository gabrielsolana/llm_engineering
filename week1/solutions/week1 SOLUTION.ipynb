````python
# imports

from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
import ollama
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_OLLAMA = 'openbmb/minicpm5-2b'

````

````python
# set up environment

load_dotenv()
openai = OpenAI()
````

````python

# set up alternative environment

ollama_client = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
#!ollama pull openbmb/minicpm5-2b
````

````python
# here is the question; type over this to ask something new

question = """
¿Qué diferencia hay entre un SW libre y otro propietario?

"""
````

````python
# prompts

system_prompt = "You are a helpful technical tutor who answers questions in English about operting systems to fresh students with litle experience.Se breve, didáctico y conciso, no respondas en más de 5-6 líneas y pon ejemplos"
user_prompt = "Please give a brief explanation to the following question: " + question
````

````python
# messages

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
````

````python
# Get ollama to answer, with streaming

stream = ollama_client.chat.completions.create(model=MODEL_OLLAMA, messages=messages,stream=True)
    
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    response = response.replace("```","").replace("markdown", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
def ask(question, client, model):
    """Stream a concise answer to a technical question."""
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "Please give a brief explanation to the following question: " + question}
    ]
    stream = client.chat.completions.create(model=model, messages=messages, stream=True)

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
# Congratulations!

my_question = input("Please enter your question:")
ask(my_question, ollama_client, MODEL_OLLAMA)
````
